# Privacy Framework - Getting Started

Welcome! This notebook walks you through setting up and using the Privacy Data Removal Framework.

## What This Framework Does

- **Discovers** your digital footprint across 50+ vectors
- **Analyzes** exposures and assigns risk scores
- **Removes** data from 9 major data brokers
- **De-indexes** URLs from Google and Bing
- **Monitors** continuously for new exposures
- **Hardens** privacy settings across accounts

## Step 1: Setup Your Identity

First, let's create your identity profile that will be used for all operations.

```python
import json
from pathlib import Path
from datetime import datetime

# Your identity information
identity = {
    "name": "Your Full Name",  # EDIT THIS
    "email": "your@email.com",  # EDIT THIS
    "phone": "+1-XXX-XXX-XXXX",  # EDIT THIS
    "github_username": "your-github-user",  # EDIT THIS
    "domain": "example.com",  # EDIT THIS
    "created_at": datetime.now().isoformat(),
    "notes": "Primary identity for privacy remediation"
}

# Save identity profile
identity_file = Path("../intel/identity_profile.json")
identity_file.parent.mkdir(parents=True, exist_ok=True)
with open(identity_file, 'w') as f:
    json.dump(identity, f, indent=2)

print("✓ Identity profile created")
print(f"\nYour Profile:")
print(json.dumps(identity, indent=2))
```

## Step 2: Check for Data Breaches

Let's check if your email appears in any known data breaches.

```python
import sys
sys.path.insert(0, '../scripts')

from hibp_monitor import HIBPMonitor

monitor = HIBPMonitor()
print(f"🔍 Checking {identity['email']} for breaches...\n")

# Check email for breaches
breach_report = monitor.check_email_breaches(identity['email'])

print(f"Email: {breach_report['email']}")
print(f"Breached: {breach_report['breached']}")
print(f"Breach Count: {breach_report['breach_count']}")
print(f"Risk Level: {breach_report['risk_level']}")

if breach_report['breached']:
    print(f"\n⚠️ ALERT: Your email was found in {breach_report['breach_count']} breach(es)\n")
    for breach in breach_report['breaches']:
        print(f"  - {breach['name']} ({breach['breach_date']})")
else:
    print(f"\n✓ Good news: No breaches detected for this email")
```

## Step 3: Discover Exposures

Now let's scan for exposures across multiple vectors.

```python
from data_broker_automation import DataBrokerAutomation

automation = DataBrokerAutomation(identity)

print("🔎 Scanning all data brokers...\n")
print("Brokers to check (9 total):")
print()

for broker_id, broker_info in automation.BROKERS.items():
    name = broker_info['name']
    difficulty = broker_info['difficulty']
    days = broker_info['time_estimate_days']
    print(f"  ✓ {name:20} ({difficulty:8} - {days:2} days)")

print(f"\nTotal: {len(automation.BROKERS)} data brokers")
print("\nTo check each broker:")
print("  1. Visit the broker's website")
print("  2. Search for your name")
print("  3. Note if you appear in results")
print("  4. Use removal_url to submit opt-out request")
```

## Step 4: Generate Removal Plan

Create a phased removal plan optimized by difficulty.

```python
print("📋 Generating 3-phase removal plan...\n")

plan = automation.generate_removal_plan(prioritize_by='difficulty')

print(f"Total Brokers: {plan['total_brokers']}")
print(f"Estimated Total Time: {plan['total_estimated_hours']:.1f} hours")
print(f"\n3-Phase Removal Strategy:\n")

for phase in plan['phases']:
    print(f"Phase {phase['phase']}: {phase['name']}")
    print(f"  Duration: {phase['duration_days']} days")
    print(f"  Brokers: {len(phase['brokers'])}")
    for broker in phase['brokers']:
        print(f"    - {broker['name']} ({broker['time_estimate_hours']:.0f} hours)")
    print()

# Save plan
plan_file = Path("../logs/removal_plan_generated.json")
plan_file.parent.mkdir(parents=True, exist_ok=True)
with open(plan_file, 'w') as f:
    json.dump(plan, f, indent=2)
print(f"✓ Plan saved to logs/removal_plan_generated.json")
```

## Step 5: De-index from Search Engines

Set up Google and Bing removal requests.

```python
from gsc_removal_agent import GSCRemovalAgent
from bing_removal_agent import BingRemovalAgent

# URLs to remove
urls_to_remove = [
    f"https://github.com/{identity['github_username']}",
    f"https://github.com/{identity['github_username']}?tab=repositories",
]

print("🔍 Setting up search engine removal...\n")

# Google removal plan
gsc = GSCRemovalAgent()
gsc_plan = gsc.generate_removal_plan(urls_to_remove, priority='high')
print(f"Google: {len(urls_to_remove)} URLs scheduled for removal")
print(f"  Timeline: {gsc_plan['timeline']['completion']}")

# Bing removal plan
bing = BingRemovalAgent()
bing_plan = bing.generate_removal_batch(urls_to_remove)
print(f"\nBing: {bing_plan['total_urls']} URLs scheduled for removal")
print(f"  Timeline: {bing_plan['timeline']['completion']}")

print(f"\n✓ Search removal plans generated")
```

## Step 6: Set Up Monitoring

Configure continuous monitoring for your identity.

```python
from monitoring_orchestrator import MonitoringOrchestrator

orchestrator = MonitoringOrchestrator(identity)

print("📊 Setting up monitoring infrastructure...\n")

# Setup all monitoring jobs
daily = orchestrator.setup_daily_checks()
weekly = orchestrator.setup_weekly_checks()
monthly = orchestrator.setup_monthly_checks()

print(f"Daily Checks: {len(daily)}")
for job in daily:
    print(f"  - {job['name']} ({job['frequency']})")

print(f"\nWeekly Checks: {len(weekly)}")
for job in weekly:
    print(f"  - {job['name']} ({job['frequency']})")

print(f"\nMonthly Checks: {len(monthly)}")
for job in monthly:
    print(f"  - {job['name']} ({job['frequency']})")

print(f"\n✓ Total monitoring jobs: {len(orchestrator.monitoring_jobs)}")
print(f"  Coverage: Continuous 24/7 monitoring active")
```
