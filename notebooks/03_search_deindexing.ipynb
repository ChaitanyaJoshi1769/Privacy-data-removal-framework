## How Search Indexing Works

### The Crawl → Index → Serve Pipeline

```
Data Broker Website
       ↓ (crawler visits)
Crawler Downloads Page
       ↓ (analyzes content)
Search Engine Indexes Page
       ↓ (stores in database)
Search Results
       ↓ (appears when searching)
User Finds Your Info
```

### Removal Strategies

**Strategy 1: URL Removal Request**
- Submit specific URL for removal
- Search engine stops crawling that URL
- Removes from index within days
- Best for: Known specific URLs

**Strategy 2: robots.txt Blocking**
- Tell crawlers not to index
- Prevents future indexing
- Removes from index over time
- Best for: Preventing future crawling

**Strategy 3: Cache Removal**
- Remove cached version
- Users can't see old version
- Cached page still appears in results
- Best for: Outdated information

**Strategy 4: Content Removal**
- Remove page from source
- Search engine eventually de-indexes
- Most permanent solution
- Best for: Pages you control
